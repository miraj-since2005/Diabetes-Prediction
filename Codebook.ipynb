# EDA

```python
corr = df.corr(numeric_only=True)
sns.heatmap(
    corr,
    annot=True,
)

plt.tight_layout()
plt.show()
```

```python
cat_cols = ['gender', 'smoking_history']

for cols in cat_cols:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=cols, hue='diabetes')
    plt.tight_layout()
```

```python
cols = ['age','bmi', 'HbA1c_level', 'blood_glucose_level']

for col in cols:
    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df, x='diabetes', y=col)
    plt.tight_layout()
    plt.show()
```

# DATA PREPROCESSING

```python
X = df.drop('diabetes', axis=1)
y = df['diabetes']
```

```python
cat_cols = ['gender', 'smoking_history']

X = pd.get_dummies(X, columns=cat_cols, drop_first=True, dtype=int)    
```

```python
X.head()
```

```python
from sklearn.preprocessing import StandardScaler

cols = ['age', 'bmi', 'blood_glucose_level', 'HbA1c_level']
for i in cols:
    scale = StandardScaler()
    X[i] = scale.fit_transform(X[cols])
```
